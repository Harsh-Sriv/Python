```python
from pydantic import BaseModel

class  user(BaseModel):
    name: str
    age: int
    email: str

agent = Agent(
    'google:gemini-3.6-flash',
    system_prompt="You are a helpful assistant that creates a user object based on the provided information.",
    output_type=user
)

response = await agent.run('Harsh is 22 year old and his email is harsh@example.com')
print(response.output)
```

```python
type(response.output)  # <class '__main__.user'>
```

```python
agent=Agent(
    'google:gemini-3.6-flash',
    system_prompt="You are a helpful assistant that provides accurate answers"
)

async with agent.run_stream("What is Pydantic AI?") as run:
    async for output in run.stream_text():
        print(output)
```

```python
agent=Agent(
    'google:gemini-3.6-flash',
    system_prompt="You are a helpful assistant that provides accurate answers in a few lines"
)

async with agent.run_stream("What is Pydantic AI?") as run:
    async for chunk in run.stream_text(delta=True):
        print(chunk, end='', flush=True)
```

```python
#Instructions vs System Prompt

# System Prompt: This is a general instruction that sets the overall behavior and tone of the AI.
# Instructions: These are specific, task-oriented commands that guide the AI in performing particular actions.

from pydantic_ai import Agent

agent1 = Agent(
    'google:gemini-3.6-flash',
    instructions="If asked what is your favorite color, respond exaclty with 'blue' otherwise answer normally."
)

result1 = await agent1.run("What is 2+2")
history = result1.all_messages()

agent2 = Agent(
    'google:gemini-3.6-flash',
    instructions="If asked what is your favorite color, respond exactly with 'red' otherwise answer normally."
)

result2 = await agent2.run("What is your favorite color?", message_history=history)
print(result2.output)  # The output will be 'red' 
```

```python
agent=Agent(
    'google:gemini-3.6-flash',
    system_prompt="You are a helpful assistant that provides accurate answers in a few lines"
)

result = await agent.run("What is my favourite color?")
print(result.output)#so right now the agent dosent have any previous context of the conversation. It will answer normally.
```
