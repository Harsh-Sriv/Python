```python
from pydantic_ai import Agent
```

```python
agent = Agent(
    'google:gemini-3.6-flash',
    system_prompt="You are a helpful assistant that provides accurate and concise answers in a single line.",
)
```

```python
response = await agent.run("What is Pydantic AI?")

print(response.output)
```

```python
from pydantic import BaseModel

class  user(BaseModel):
    name: str
    age: int
    email: str

agent = Agent(
    'google:gemini-3.6-flash',
    system_prompt="You are a helpful assistant that creates a user object based on the provided information.",
    output_type=user
)

response = await agent.run('Harsh is 22 year old and his email is harsh@example.com')
print(response.output)
```

```python
type(response.output)  # <class '__main__.user'>
```

```python
agent=Agent(
    'google:gemini-3.6-flash',
    system_prompt="You are a helpful assistant that provides accurate answers in a single line."
)

async with agent.run_stream("What is Pydantic AI?") as run:
    async for output in run.stream_text():
        print(output)
```
